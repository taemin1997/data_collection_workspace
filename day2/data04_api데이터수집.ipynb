```python
import requests
```

```python
url = 'https://apis.data.go.kr/B552061/frequentzoneBicycle/getRestFrequentzoneBicycle?ServiceKey=#&searchYearCd=2021&siDo=11&guGun=650&type=json&numOfRows=10&pageNo=1'
response = requests.get(url)
print(response)            #<Response [200]>
print(dir(response))
print(type(response))     # requests.models.Response
print(response.json())

```

```python
# item 접근
w_data = response.json()['items']['item']
print(type(w_data))               # list
print((len(w_data)))              # 갯수


for row in range(len(w_data)):
  print('spot_nm: ', w_data[row]['spot_nm'])
  print('경도: ', w_data[row]['lo_crd'])
  print('위도: ', w_data[row]['la_crd'])


result = []
data = response.json()
w_data = data['items']['item']
for row in range(len(w_data)):
  line = {}
  line['spot_nm'] = w_data[row]['spot_nm']
  line['lo_crd'] = w_data[row]['lo_crd']
  line['la_crd'] = w_data[row]['la_crd']
  result.append(line)
```

```python
result
```

```python
import json
```

```python
with open('./cycle.json', 'w', encoding = 'utf-8') as f :
    json.dump(result, f, indent='\t', ensure_ascii=False)
```

```python
with open('./cycle.json', 'r', encoding = 'utf-8') as f :
    data = json.load(f)
    print(data)
```

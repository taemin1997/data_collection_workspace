```python
html_str = """<!DOCTYPE html>
<html>
  <body>
    <ul class="ko">
       <li>
           <p> Previous Element1</p>
           <p> Previous Element2</p>
           <a  href ="https://www.naver.com"> 네이버</a>
           <p> Next Element1</p>
           <p> Next Element2</p>
       </li>


        <li>
           <a  href ="https://www.daum.net"> 다음</a>
       </li>
    </ul>


    <ul class="sns">
       <li>
          <a href="https://www.google.com">구글</a>
       </li>
       <li>
          <a href="https://www.facebook.com">페이스북</a>
       </li>
    </ul>
    <div id="result">page</div>
    <p>


  </body>
</html>
"""
```

```python
import bs4

bs_obj = bs4.BeautifulSoup(html_str, 'html.parser')
```

```python
print(type(bs_obj))
```

```python
print(bs_obj.find_all('div'))

print(bs_obj.find('div').text)
```

```python
ulEle = bs_obj.find('ul')
print(ulEle)
```

```python
ul_li = ulEle.find('li')
print(ul_li)
```

```python
print(ul_li.text)
```

```python
ul = bs_obj.find('ul')
ul_li = ul.find_all('li')

print(ul_li[1])
```

```python
for e in ul_li : 
    print(e.text)
    print('---')
```

```python
ul_ko = bs_obj.find('ul', {'class' : 'ko'})
print(ul_ko)
```

```python
ul_ko = bs_obj.find('ul', class_ = 'ko')
print(ul_ko)
```

```python
ul_sns = bs_obj.find('ul', attrs = {'class' : 'sns'})
print(ul_sns)
```

```python
ul_li = bs_obj.find('ul', attrs = {'class' : 'sns'}).find_all('li')
print(ul_li)
```

```python
for i in ul_li :
    print(i.text)
```

```python
import re

print(bs_obj.find('a', {'href':'https://www.naver.com'}))
print(bs_obj.find('a', {'href': re.compile('com$')}))              # com으로 끝나는
print(bs_obj.find('a', {'href': re.compile('.*naver.*')}))         # naver가 포함된
print(bs_obj.find('a', {'href': re.compile('naver')}))             # naver가 포함된
print(bs_obj.find('a', {'href': re.compile('^https')}))            # https으로 시작하는

print('-------------------------------------------')
print(bs_obj.find_all('a', {'href':'https://www.naver.com'}))
print(bs_obj.find_all('a', {'href': re.compile('com$')}))              # com으로 끝나는
print(bs_obj.find_all('a', {'href': re.compile('.*naver.*')}))         # naver가 포함된
print(bs_obj.find_all('a', {'href': re.compile('naver')}))             # naver가 포함된
print(bs_obj.find_all('a', {'href': re.compile('^https')}))            # https으로 시작하는

print('-------------------------------')
print(bs_obj.find(href='https://www.naver.com'))
print(bs_obj.find(href=lambda val : val))
print(bs_obj.find(href=lambda val : val and 'daum' in val))          # daum이 포함될때가지 검색을 함


print(bs_obj.find(class_=lambda val : val and 'sns' in val))        # class속성의 값이 sns가 포함되여 일때 가져온다.

```
